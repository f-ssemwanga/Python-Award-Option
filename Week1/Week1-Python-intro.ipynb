### Concept 1: Variables and data types

A variable is a name that holds a value. Python works out the type for you, but you still need to know what it picked, because the type decides what operations are allowed.

| Type | What it holds | Example |
| --- | --- | --- |
| `int` | Whole numbers | `5`, `-3`, `2026` |
| `float` | Numbers with a decimal point | `3.14`, `-0.001`, `17.5` |
| `str` | Text, in quotes | `"hello"`, `"123"` |
| `bool` | True or False | `True`, `False` |

**Before you run the cell below, predict what the last line will print.**

````python
station = "Oxford"
year = 2025
mean_max_temp = 15.2
is_in_england = True

print(station, type(station))
print(year, type(year))
print(mean_max_temp, type(mean_max_temp))
print(is_in_england, type(is_in_england))

print("Type of year + mean_max_temp:", type(year + mean_max_temp))
````

**Now break it.** Change `year = 2025` to `year = "2025"`, with quotes, and run the cell again.

You should get a `TypeError`. Read the message rather than skipping past it. Python is telling you it will not add a string to a float, and the error names both types. Learning to read errors now will save you hours later in the course.

### Concept 2: Input and output

`input()` asks the user for something. `print()` displays something.

The single most important thing on this page: **`input()` always gives you a string**, even when the user types a number. If you want to do arithmetic with it, you must convert it first.

- `int("42")` gives the integer `42`
- `float("17.5")` gives the float `17.5`

Run the cell below and type a number when it asks.

````python
rainfall_text = input("Enter a rainfall figure in mm: ")

print("You typed:", rainfall_text)
print("Python thinks that is a:", type(rainfall_text))

rainfall = float(rainfall_text)
print("After converting:", type(rainfall))
print("Doubled:", rainfall * 2)
````

**Now break it.** Delete the `float(...)` conversion and use `rainfall_text * 2` instead.

You will not get an error. You will get the number printed twice, side by side. This is worse than an error, because the program carries on and gives you a wrong answer silently. Most real bugs look like this.

### Concept 3: Arithmetic

| Operator | Does | `7` and `2` give |
| --- | --- | --- |
| `+` | Add | `9` |
| `-` | Subtract | `5` |
| `*` | Multiply | `14` |
| `/` | Divide, always a float | `3.5` |
| `//` | Integer division, throws away the remainder | `3` |
| `%` | Modulus, the remainder only | `1` |
| `**` | To the power of | `49` |

`//` and `%` are the two that catch people out, and they are genuinely useful. `%` is how you test whether a number is even, and how you get the last digit of a number.

````python
a = 7
b = 2

print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} / {b}  = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b}  = {a % b}")
print(f"{a} ** {b} = {a ** b}")
````

### Concept 4: f-strings

An f-string lets you drop a variable straight into a piece of text. Put `f` before the opening quote, then put the variable inside curly brackets.

```python
name = "Sam"
print(f"Hello, {name}")
```

You can also control how numbers are displayed, which matters as soon as you are dealing with real measurements:

| Format | Means | `15.23456` becomes |
| --- | --- | --- |
| `{value:.1f}` | One decimal place | `15.2` |
| `{value:.2f}` | Two decimal places | `15.23` |
| `{value:>8}` | Pad to 8 characters, right aligned | `15.23456` |

````python
station = "Oxford"
month = "July"
year = 2026
tmax = 28.2
rain = 0.0
sun = 353.0

print(f"{station}, {month} {year}")
print(f"  Mean daily maximum: {tmax:.1f} C")
print(f"  Total rainfall:     {rain:.1f} mm")
print(f"  Sunshine:           {sun:.1f} hours")
````

Those are the real Oxford readings for July 2026. No rain at all that month, and 353 hours of sunshine.

**Now break it.** Change `{tmax:.1f}` to `{tmax:.3f}` and run it again. Python will happily print `28.200`, showing three decimal places of precision that the measurement does not actually have. Formatting can imply accuracy that is not there, which is worth remembering when you present results.
