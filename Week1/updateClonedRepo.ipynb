### Getting each new week's material

Your fork is a copy of the course taken on the day you forked it. When your teacher adds next week's notebooks to the course, your fork does **not** update by itself. You pull the new material in, in two steps: first into your fork on GitHub, then down to this computer.

Do this at the **start of every lesson**, before you change anything.

### Step 1: Save and sync your own work first

In VS Code, save, then commit and sync anything you have changed (`usingGitHub.ipynb`, Part 5).

**You should see:** nothing listed under **Changes** in the Source Control panel.

Doing this first means your work is safe on GitHub before anything new arrives.

### Step 2: Sync your fork on GitHub

1. Go to your fork on github.com: `github.com/your-username/Python-Award-Option`
2. Just above the list of files there is a line saying your branch is some number of *commits behind* f-ssemwanga/Python-Award-Option. Next to it, click **Sync fork**, then **Update branch**.

**You should see:** *This branch is up to date with f-ssemwanga/Python-Award-Option:main.*

If it already said *up to date* before you clicked anything, there is nothing new. Go to Step 3 anyway, it does no harm.

**Stop if you see *This branch has conflicts*, and a red *Discard commits* button.** Do **not** click Discard. It deletes your own work from GitHub to make room for the teacher's version. Tell your teacher instead. This only happens if you and your teacher have both changed the same file, and it takes a minute to sort out properly.

### Step 3: Pull it down to this computer

In VS Code, open the Source Control panel, click the **...** at the top, and choose **Pull**. Or click the circular arrows next to `main` at the bottom left.

**You should see:** the new week's folder appear in the left hand panel.

### The same thing from the terminal

Useful if you want to see what is happening, or if the Sync fork button is not there.

Check which remote repositories your copy knows about:

```bash
git remote -v
```

**You should see** two lines for `origin`, which is your fork. If there are no lines for `upstream`, which is the teacher's course, add it. You only ever do this once per clone:

```bash
git remote add upstream https://github.com/f-ssemwanga/Python-Award-Option.git
```

Then, every time you want the new material:

```bash
git fetch upstream
git merge upstream/main
git push origin main
```

`fetch` downloads what is new without touching your files. `merge` brings it into your copy. `push` sends the result up to your fork, so GitHub and this computer match.

**If `merge` reports a conflict**, VS Code highlights the file in the Source Control panel. For a `.py` or `.md` file you can pick which version to keep. For a notebook, ask your teacher, because conflicts inside `.ipynb` files are much harder to read.
