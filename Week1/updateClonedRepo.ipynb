### Getting each new week's material

Your fork is a copy of the course taken on the day you forked it. When your teacher adds next week's notebooks to the course, your fork does **not** update by itself. You pull the new material in, in two steps: first into your fork on GitHub, then down to this computer.

Do this at the **start of every lesson**, before you change anything.

### Step 1: Save and sync your own work first

In VS Code, save, then commit and sync anything you have changed (`usingGitHub.ipynb`, Part 5).

**You should see:** nothing listed under **Changes** in the Source Control panel.

Doing this first means your work is safe on GitHub before anything new arrives.

### Step 2: Sync your fork on GitHub

1. Go to your fork on github.com: `github.com/your-username/Python-Award-Option`
2. Just above the list of files there is a line saying your branch is some number of *commits behind* f-ssemwanga/Python-Award-Option. Next to it, click **Sync fork**, then **Update branch**.

**You should see:** *This branch is up to date with f-ssemwanga/Python-Award-Option:main.*

If it already said *up to date* before you clicked anything, there is nothing new. Go to Step 3 anyway, it does no harm.

**Stop if you see *This branch has conflicts*, and a red *Discard commits* button.** Do **not** click Discard. It deletes your own work from GitHub to make room for the teacher's version. Tell your teacher instead. This only happens if you and your teacher have both changed the same file, and it takes a minute to sort out properly.

### Step 3: Pull it down to this computer

In VS Code, open the Source Control panel, click the **...** at the top, and choose **Pull**. Or click the circular arrows next to `main` at the bottom left.

**You should see:** the new week's folder appear in the left hand panel.

### The same thing from the terminal

Useful if you want to see what is happening, or if the Sync fork button is not there.

Check which remote repositories your copy knows about:

```bash
git remote -v
```

**You should see** two lines for `origin`, which is your fork. If there are no lines for `upstream`, which is the teacher's course, add it. You only ever do this once per clone:

```bash
git remote add upstream https://github.com/f-ssemwanga/Python-Award-Option.git
```

Then, every time you want the new material:

```bash
git fetch upstream
git merge upstream/main --no-edit
git push origin main
```

`fetch` downloads what is new without touching your files. `merge` brings it into your copy. `--no-edit` stops Git opening an editor to ask for a message; without it Git Bash drops you into Vim, and you get out by typing `:q` and pressing Enter. `push` sends the result up to your fork, so GitHub and this computer match.

**If `merge` reports a conflict**, VS Code highlights the file in the Source Control panel. For a `.py` or `.md` file you can pick which version to keep. For a notebook, ask your teacher, because conflicts inside `.ipynb` files are much harder to read.

### Never committed your work, or never forked?

Use this if your work has only ever been saved on this computer. Open a terminal in VS Code (**Terminal**, **New Terminal**). It opens in the course folder. Run each command, check the result, then move on.

**1. Where does your copy point?**

```bash
git remote -v
```

- If the `origin` lines show **your own username**, you have a fork. If there are no `upstream` lines, add them:

```bash
git remote add upstream https://github.com/f-ssemwanga/Python-Award-Option.git
```

- If the `origin` lines show **f-ssemwanga**, you cloned the teacher's course and have nowhere of your own to save to. First go to `github.com/f-ssemwanga/Python-Award-Option`, click **Fork**, then **Create fork**. Then run these two lines, putting your GitHub username in the second:

```bash
git remote rename origin upstream
git remote add origin https://github.com/YOUR-USERNAME/Python-Award-Option.git
```

**You should see**, when you run `git remote -v` again: `origin` with your username and `upstream` with f-ssemwanga.

**2. Check Git knows who you are**

```bash
git config user.name
```

**You should see** your name. If nothing prints, set it as in `usingGitHub.ipynb` Part 4.

**3. Commit everything you have done so far**

```bash
git add -A
git commit -m "My work up to Week 6"
```

**You should see** a line saying how many files changed. *Nothing to commit* is fine too: it means it was already done.

**4. Bring in the new material**

```bash
git fetch upstream
git merge upstream/main --no-edit
```

**If it says CONFLICT**, you and your teacher both changed the same notebook. Keep your own copy:

```bash
git checkout --ours -- .
git add -A
git commit --no-edit
```

**5. Send it all to your GitHub**

```bash
git push origin main
```

The first time, a window asks you to sign in to GitHub. **You should see** a line ending in `main -> main`. Check your fork on github.com: your notebooks are there.

### If Pull stops with a conflict in a Week 1 to 4 notebook

The Week 1 to 4 notebooks were tidied in November 2026: the challenges are now split into a Core and a Fast Track notebook. If you had already typed into the old ones, Pull may stop and say there is a conflict.

Keep your own copy. Nothing new is lost, because the model answers are in each week's `Solutions` notebook. In the terminal, run these three lines:

```bash
git checkout --ours -- Week1 Week2 Week3 Week4
git add Week1 Week2 Week3 Week4
git commit --no-edit
```

**You should see:** a line ending in `Merge branch 'main'`, and nothing listed under **Changes** in the Source Control panel. Your old notebooks are kept, and the new Core and Fast Track ones sit beside them.
