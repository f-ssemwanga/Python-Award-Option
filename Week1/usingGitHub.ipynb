### Week 1: Your own copy of the course on GitHub

**Time:** 20 minutes, taught to the whole class. Both lanes do this.

By the end you will have your own copy of the course on GitHub, a copy of it on this computer, and your first piece of work saved to both.

#### Why bother

- Your work is saved somewhere you cannot lose it, whichever computer you use next lesson.
- Every version you save is kept. If you break something, you can go back.
- The same account holds your portfolio later in the course (`Portfolio_Guidance.ipynb`), which is what a university or employer will actually look at.

#### Three words to know

| Word | What it means here |
| --- | --- |
| **Fork** | Your own copy of the course, on the GitHub website |
| **Clone** | A copy of your fork on the computer in front of you |
| **Commit and Sync** | Save a snapshot of your work, then send it up to your fork |

You fork once, clone once per computer, and commit and sync every lesson.

### Part 1: Create a GitHub account

Skip this if you already have one.

1. Go to [github.com](https://github.com) and click **Sign up**.
2. Pick a **professional username**. It will be in the link you put on your CV and UCAS statement. `j-smith-dev` is good. `xXgamer99Xx` is not.
3. Verify your email address.

**You should see:** your GitHub home page, signed in.

### Part 2: Fork the course

1. Go to the course repository: [https://github.com/f-ssemwanga/Python-Award-Option](https://github.com/f-ssemwanga/Python-Award-Option)
2. Click **Fork**, near the top right.
3. Leave everything as it is, including the tick in **Copy the main branch only**, and click **Create fork**.

**You should see:** a page titled **your-username / Python-Award-Option**, with *forked from f-ssemwanga/Python-Award-Option* underneath.

This is yours. You can change anything in it and nothing happens to the teacher's copy.

### Part 3: Clone your fork in VS Code

1. On your fork's page, click the green **Code** button, make sure **HTTPS** is selected, and click the copy icon next to the link. It ends in `.git` and has **your** username in it, not `f-ssemwanga`.
2. In VS Code, close any folder that is open: **File**, then **Close Folder**.
3. Click the **Source Control** icon in the left bar (three dots joined by lines), or press Ctrl, Shift and G.
4. Click **Clone Repository**, paste the link, and press Enter.
5. Choose where to put it. **Documents** is a good place. Avoid Downloads and the Desktop.
6. When VS Code asks *Would you like to open the cloned repository?*, click **Open**. If it asks whether you trust the authors, click **Yes**.

**You should see:** `Week1`, `Week2` and the rest in the left hand panel, and `main` at the bottom left of the window.

**If VS Code says Git is not installed**, stop and tell your teacher. Do not download the ZIP instead: a ZIP copy cannot sync, so your work would only exist on this one machine.

### Part 4: Tell Git who you are

Git labels every snapshot with a name and email. You set this once for each clone.

Open a terminal inside VS Code: **Terminal**, then **New Terminal**. Type these two lines, one at a time, with your own details:

```bash
git config user.name "Your Name"
git config user.email "the-email-you-used-for-github@example.com"
```

**You should see:** nothing at all. No message means it worked. To check, type `git config user.name` and it prints your name back.

**Why there is no `--global` here.** Online guides tell you to add `--global`. On a shared school computer that would stamp your name on the next person's work as well. Without it, the setting belongs to your copy of the course only. At home, on your own machine, `--global` is fine.

### Part 5: Your first commit and sync

1. Open `week1Challenges.ipynb`, type anything into the first code cell, and save with **Ctrl and S**.
2. Click the **Source Control** icon. The notebook is listed under **Changes** with an **M** next to it, for modified.
3. In the **Message** box at the top, type what you did, for example `Start Week 1 challenges`.
4. Click **Commit**. If VS Code asks whether to stage all your changes and commit them, click **Yes**.
5. Click **Sync Changes**. The first time, a browser window asks you to sign in to GitHub and authorise VS Code. Do that, then come back.

**You should see:** on your fork's page on github.com, your commit message next to `Week1`, and *now* or *1 minute ago* next to it.

A commit message says what you did, not that you did something. `Finish Core 3 name details` is useful in three months' time. `update` is not.

#### The same thing from the terminal

VS Code's buttons are running these commands for you. You do not need them, but it helps to know what is happening underneath.

| Command | Does |
| --- | --- |
| `git status` | Shows what has changed since the last commit |
| `git add .` | Stages every change, ready to commit |
| `git commit -m "Finish Core 3"` | Takes the snapshot, with a message |
| `git push` | Sends your commits up to GitHub |
| `git pull` | Brings down commits from GitHub that you do not have |

### Appendix A: SSH keys, for your own computer only

Not needed in lessons. The HTTPS sign in from Part 5 is enough. SSH saves you signing in on a machine you use every day. Do not set this up on a shared school computer.

1. Check whether you already have a key:

```bash
ls -al ~/.ssh
```

If you see `id_ed25519` and `id_ed25519.pub`, skip to step 3.

2. Make a new key, using your GitHub email. Press Enter to accept the default location, then choose a passphrase:

```bash
ssh-keygen -t ed25519 -C "your_email@example.com"
```

3. Start the SSH agent and add the key:

```bash
eval "$(ssh-agent -s)"
ssh-add ~/.ssh/id_ed25519
```

4. Copy the **public** key, the one ending `.pub`. Never share the other file.

```bash
cat ~/.ssh/id_ed25519.pub | clip      # Windows, in Git Bash
cat ~/.ssh/id_ed25519.pub | pbcopy    # Mac
```

5. On GitHub: your profile picture, **Settings**, **SSH and GPG keys**, **New SSH key**. Give it a title such as *My laptop*, paste the key, click **Add SSH key**.

6. Test it:

```bash
ssh -T git@github.com
```

**You should see:** `Hi your-username! You've successfully authenticated, but GitHub does not provide shell access.` That message is a success, despite how it reads.

To clone with SSH, copy the **SSH** link from the green Code button instead of the HTTPS one. It looks like `git@github.com:your-username/Python-Award-Option.git`.

### Appendix B: Git on a Raspberry Pi

In the terminal, update the system, then install Git:

```bash
sudo apt update
sudo apt upgrade
sudo apt install git
```

Check it installed:

```bash
git --version
```

**You should see:** something like `git version 2.39.5`.

Tell Git who you are. A Pi is usually your own machine, so `--global` is fine here:

```bash
git config --global user.name "Your Name"
git config --global user.email "your_email@example.com"
git config --list
```

The last line prints your settings so you can check them.

### Appendix C: Working on someone else's repository with branches

For later in the course, when you work on a repository with other people. You will not need it in Week 1.

When several people push to the same repository, work on a **branch** so you are not all changing `main` at once.

1. Get up to date, then make a branch and switch to it:

```bash
git pull origin main
git switch -c feedback
```

2. Make your changes, then commit them as normal:

```bash
git add .
git commit -m "Describe what you changed"
```

3. Push the branch so others can see it:

```bash
git push -u origin feedback
```

4. On GitHub, open a **pull request** from `feedback` into `main`. The others review it there and merge it when it is ready. This is how almost every team works.

5. Once it is merged, tidy up:

```bash
git switch main
git pull origin main
git branch -d feedback
git push origin --delete feedback
```
