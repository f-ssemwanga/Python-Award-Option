## Core

### Core 1: Temperature converter

```python
celsius = float(input("Enter a temperature in Celsius: "))
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius:.1f} C is {fahrenheit:.1f} F")
```

**Why:** `input()` always gives back a string, even if you type `21`. `float()` turns it into a number so the sum works. Without it, Python tries to multiply text by 9 and crashes. `:.1f` gives one decimal place.

### Core 2: House points report

```python
house = "Rowe"
term = "Autumn 2026"
academic = 412
sport = 358
music = 97
pupils = 64

total = academic + sport + music
per_pupil = total / pupils

print(f"House report: {house}, {term}")
print(f"  Academic points: {academic:>4}")
print(f"  Sport points:    {sport:>4}")
print(f"  Music points:    {music:>4}")
print(f"  Total points:    {total:>4}")
print(f"  Points per pupil: {per_pupil:.1f}")
```

**Why:** work the numbers out first, then print. Keeping the calculation apart from the printing means you can change the figures at the top and the whole report updates. `:>4` right-aligns each number in four spaces, so the column lines up.

### Core 3 (extension): Bill splitter

```python
total = float(input("Enter the bill total in pounds: "))
total_pence = round(total * 100)

each_pence = total_pence // 3
left_over = total_pence % 3

print(f"Each person pays £{each_pence / 100:.2f}")
print(f"{left_over}p is left over")
```

**Why:** money is safer in whole pence. `//` gives the whole number of pence each person pays and `%` gives what is left. Use `round()`, not `int()`: inside the computer `19.99 * 100` is `1998.9999999999998`, and `int()` would chop it to 1998.

`100` gives £33.33 and 1p left over. `19.99` gives £6.66 and 1p left over.

## Fast Track

### Fast Track 1: Two more functions

```python
def celsius_to_fahrenheit(celsius):
    """Convert a temperature from Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32


def fahrenheit_to_celsius(fahrenheit):
    """Convert a temperature from Fahrenheit to Celsius."""
    return (fahrenheit - 32) * 5 / 9


def temperature_range(tmax, tmin):
    """Return the difference between the highest and lowest temperature."""
    return round(tmax - tmin, 1)


print(fahrenheit_to_celsius(212))
print(temperature_range(7.2, 2.1))
```

**Why:** the brackets in `(fahrenheit - 32) * 5 / 9` matter: without them Python does `32 * 5 / 9` first. In `temperature_range`, `7.2 - 2.1` is really `5.1000000000000005` because of how computers store decimals, so `round(..., 1)` tidies it to `5.1`.

### Fast Track 2: Initials

```python
def initials(full_name):
    """Return the initials of a name in capitals, each followed by a full stop."""
    result = ""
    for word in full_name.split():
        result = result + word[0].upper() + "."
    return result


print(initials("francis ssemwanga"))
print(initials("Mary Jane Watson"))
print(initials("  Ada   Lovelace "))
```

**Why:** `split()` with nothing in the brackets splits on any run of spaces and ignores spaces at the ends, which is what makes the third test pass. `split(" ")` gives empty strings for the extra spaces, and `word[0]` crashes on an empty string.

A shorter version, once you meet joins: `return "".join(word[0].upper() + "." for word in full_name.split())`

### Fast Track 3 (extension): Improve someone else's code

```python
def average_of_three(first, second, third):
    """Return the mean of three numbers."""
    return (first + second + third) / 3


print(f"Average: {average_of_three(7.2, 15.6, 28.2):.1f}")
```

**Model sentence:** renaming `f` to `average_of_three` made the biggest difference, because the call now says what it does and you no longer have to read the body to find out.
