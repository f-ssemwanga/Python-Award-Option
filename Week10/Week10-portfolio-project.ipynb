### Starter program

This runs as it is. It already has a menu, a working option 1 and a validated option 2. Your job is to build it out to meet the checklist.

Run the first cell, then the starter. Type `2`, then a station name, to see it work.

```python
# Run this cell first. It lets the notebook find the course data folder,
# whether it runs from the week folder or the course folder.
import sys; sys.path.append(".."); from coursetools import data_file
```

```python
import csv

STATIONS = ["Armagh", "Oxford", "Sheffield", "Lerwick", "Camborne"]


def load_rows():
    """Return every row of the weather CSV as a list of lists."""
    rows = []
    with open(data_file("uk_weather.csv")) as file:
        reader = csv.reader(file)
        next(reader)
        for row in reader:
            rows.append(row)
    return rows


def ask_station():
    """Keep asking until the user types one of the five station names."""
    while True:
        name = input(f"Station ({', '.join(STATIONS)}): ").strip().title()
        if name in STATIONS:
            return name
        print("That station is not in the data. Check the spelling.")


def show_menu():
    print()
    print("1. List the stations")
    print("2. Count the months of data for a station")
    print("3. Quit")


rows = load_rows()
while True:
    show_menu()
    choice = input("Choose 1, 2 or 3: ").strip()
    if choice == "1":
        for name in STATIONS:
            print(name)
    elif choice == "2":
        station = ask_station()
        count = 0
        for row in rows:
            if row[0] == station:
                count = count + 1
        print(f"{station} has {count} months of data")
    elif choice == "3":
        print("Goodbye")
        break
    else:
        print("Please type 1, 2 or 3")
```

**Where to start:**

1. Add option 3, **summary for a station and year**: total rain, warmest month, coldest month. Ask for the year with a validation loop, so `twenty` and `1066` are both rejected kindly. Move Quit to option 4.
2. Then make a `WeatherStation` class, as in Week 7, and move the summary maths into its methods.
3. Then the chart, then saving.

`', '.join(STATIONS)` joins the names into one string with commas between them. `.strip().title()` tidies what the user typed, so `  oxford` still matches `Oxford`.
