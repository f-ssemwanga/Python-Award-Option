### Week 10: Portfolio Project

#### Lesson Objectives:
- Bring together the skills from Weeks 1 to 9 into a single assessed program.
- Design and build a program that models real-world entities using classes.
- Store and retrieve data in a SQLite database from Python code.
- Validate user input and handle errors so the program does not crash.
- Commit the finished project to your own GitHub portfolio repository.

This is the assessed artefact for the course. It is one program, not a set of
exercises, and it is the piece of work you will show to a university or an
employer.

The portfolio repository itself, how to set it up and how to write it up, is
covered in full in [Portfolio_Guidance.ipynb](../Portfolio_Guidance.ipynb).
Read that first. This notebook only describes the project you are adding to it.

### The project

You are building a **records management program** for a context of your own
choosing. Some suggestions:

- a club membership register
- a library or equipment loan system
- a sports team fixture and results tracker
- a science practical results log

Whatever you choose, the program must do the same four things: model the data
with classes, store it in a database, guard against bad input, and survive
errors without crashing.

Every specification below is met by a program of roughly 150 to 250 lines. If
yours is much longer than that, you have probably taken on too much.

---

### Core specification

Build a program that meets all of the following.

1. **Model the data with a class.** Define at least one class with an
   `__init__` constructor, attributes, and at least two methods. Create objects
   from it. *(Week 7)*
2. **Store the data in SQLite.** Create a database file and a table with
   sensible field names, data types, and a primary key. Write a reusable
   connection function rather than repeating the connection code. *(Week 9)*
3. **Insert, read and update records.** Your program must be able to add a new
   record, display all records, display a filtered selection of records, and
   update an existing record. *(Week 9)*
4. **Validate every input.** No value reaches the database without being
   checked. Numbers must be numbers, required fields must not be empty, and
   values from a fixed set must be one of that set. *(Week 5)*
5. **Handle errors.** Wrap anything that can fail in `try` / `except`. Your
   program must not crash on bad input, a missing file, or a database error.
   *(Week 5)*
6. **Give the user a menu.** A loop that offers the options above and exits
   cleanly when the user asks it to. *(Weeks 2 and 3)*
7. **Commit it to your portfolio.** Push the finished `.py` file to your
   portfolio repository with a clear commit message and a written reflection in
   your README. *(Week 1 and the Portfolio Guide)*

---

### Fast Track specification

Fast Track students build the same program, meeting every point of the Core
specification, and then extend it as follows. You do not do a different project;
you do the same project to a higher standard.

1. **Use inheritance.** Define a parent class and at least one child class that
   overrides a method. Justify in a comment why the child class exists rather
   than a flag on the parent. *(Week 7)*
2. **Separate your code into modules.** Put the class definitions in one `.py`
   file and the database functions in another, and import both into a main
   program guarded by `if __name__ == "__main__":`. *(Week 3)*
3. **Write your own exception class.** Define an exception that is specific to
   your program, raise it when a business rule is broken, and catch it where the
   user can be told what went wrong. *(Week 5)*
4. **Add an export.** Let the user write the current records out to a `.csv` or
   `.json` file using the `with` statement. *(Week 6)*
5. **Choose your data structures deliberately.** Somewhere in the program, use a
   dictionary or a set where a list would have been the obvious choice, and
   write two or three sentences in your README explaining why it is the better
   structure here. *(Week 4)*

### Submission checklist

Before you say you have finished, check every line of this list.

- [ ] The program runs from start to finish without crashing.
- [ ] Every input is validated, and bad input produces a helpful message rather
      than a traceback.
- [ ] The database file is created by the program if it does not already exist.
- [ ] Adding, displaying, filtering and updating records all work.
- [ ] The code is commented, and variable names say what they hold.
- [ ] The `.py` file is committed and pushed to your portfolio repository.
- [ ] Your README links to the project and explains what it does.
- [ ] Your README contains a short reflection: what was hard, and what you would
      do differently next time.
- [ ] Fast Track only: every point of the Fast Track specification is met and
      the structural choices are justified in writing.
