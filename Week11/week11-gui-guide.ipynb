### Concept 1: A window and the event loop

Open `step1_window.py` in Thonny and press **F5**. A small empty window appears.

The last line, `app.exec()`, is the important one. Every program so far ran from top to bottom and stopped. A program with a window has to **wait**: for a click, a key press, or the window closing. `app.exec()` is that waiting, and it is called the **event loop**. When you close the window it finishes, and `Window closed` prints.

**Now break it.** Delete the `app.exec()` line and run again. The window flashes up and vanishes, because nothing makes the program wait.

### Concept 2: A button that does something

Open `step2_button.py` and run it. Type a temperature and press **Convert**.

Three new ideas, each using something you already know:

| Code | What it does | You met it in |
| --- | --- | --- |
| `class ConverterWindow(QWidget)` | your window is a class built on Qt's `QWidget` | Week 7, inheritance |
| `QLineEdit`, `QPushButton`, `QLabel` | a text box, a button and a line of text | new: they are just objects |
| `self.button.clicked.connect(self.convert)` | when the button is clicked, run the `convert` method | new: **event-driven** code |

The `convert` method uses `try` and `except` exactly as in Week 5, so typing `abc` cannot crash the app.

**Now break it.** Add brackets: `connect(self.convert())`. You get `TypeError: argument 1 has unexpected type 'NoneType'`. With brackets, Python **runs** `convert` straight away and passes on what it returns, which is `None`. Without brackets, you hand over the method itself, to be run later when the button is clicked.
