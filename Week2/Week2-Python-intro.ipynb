### Concept 1: Making decisions with `if`, `elif` and `else`

Python checks each condition from the top. The **first** one that is `True` runs, and the rest are skipped. The indented lines are the ones each test controls.

Comparisons: `==` equal, `!=` not equal, `<`, `>`, `<=`, `>=`. Join two with `and` or `or`.

**Predict what it prints, then run it.**

```python
month = "June"
tmax = 24.0

if tmax >= 22:
    print(f"{month} was hot")
elif tmax >= 15:
    print(f"{month} was warm")
elif tmax >= 8:
    print(f"{month} was mild")
else:
    print(f"{month} was cold")
```

**Now break it.** Make `tmax >= 8` the first test instead of the third. June is now "mild", with no error. **The order of the tests is part of the logic.**

### Concept 2: Repeating a set number of times with `for` and `range`

A `for` loop runs its indented lines once for each value. `range(1, 13)` gives 1 to 12: the stop number is never included. `range(10, 0, -1)` counts down from 10 to 1.

```python
for month in range(1, 13):
    print(f"Month {month}")
```

**Now break it.** Change it to `range(1, 12)`. December disappears, with no error. This is an *off-by-one error*, the most common loop bug there is.

### Concept 3: Looping over a list of values

A list is a row of values in square brackets (more in Week 4). A `for` loop can visit each value in turn. This list is Oxford's mean daily maximum for each month of 2025.

The loop keeps a **counter** and a **running total**. Both start at 0 **before** the loop.

```python
tmax_2025 = [6.8, 8.4, 13.2, 17.3, 19.6, 24.0, 25.3, 24.2, 19.2, 14.6, 12.0, 9.6]

warm_months = 0
total = 0

for temp in tmax_2025:
    total = total + temp
    if temp > 20:
        warm_months = warm_months + 1

print(f"Months above 20 C: {warm_months}")
print(f"Mean for the year: {total / 12:.1f} C")
```

**Now break it.** Move `warm_months = 0` inside the loop, just above the `if`. The count drops to 0, because it is reset every time round. Put it back.

### Concept 4: Repeating until something changes, with `while`

A `while` loop repeats as long as its condition is `True`. Use it when you cannot know in advance how many times round you need, like checking input.

Run it and type `13`, then `0`, then `6`.

```python
tmax_2025 = [6.8, 8.4, 13.2, 17.3, 19.6, 24.0, 25.3, 24.2, 19.2, 14.6, 12.0, 9.6]

month = int(input("Enter a month number, 1 to 12: "))

while month < 1 or month > 12:
    print("That is not a month.")
    month = int(input("Enter a month number, 1 to 12: "))

print(f"Month {month}: mean daily maximum {tmax_2025[month - 1]} C")
```

**Now break it.** Change `or` to `and` and type `13`. No number can be below 1 **and** above 12, so the check never runs, and the last line crashes. Put `or` back.

**`for` or `while`?** `for` when you know how many times. `while` when you stop because something happens.
