## Core

### Core 1: pH classifier

```python
ph = float(input("Enter the pH reading: "))

if ph < 3:
    label = "Strong acid"
elif ph < 7:
    label = "Weak acid"
elif ph == 7:
    label = "Neutral"
elif ph <= 11:
    label = "Weak alkali"
else:
    label = "Strong alkali"

print(f"pH {ph}: {label}")
```

**Why:** the tests run top to bottom and stop at the first one that is true. So by the time Python reaches `elif ph < 7`, it already knows the pH is not below 3. Each test only needs its upper limit.

The edges: `3` is a weak acid, `7` is neutral, `11` is a weak alkali, `11.5` is a strong alkali.

### Core 2: Radioactive decay

```python
count = 1600
hours = 0
print(f"{hours} hours: {count:.1f} counts per minute")

while count >= 50:
    count = count / 2
    hours = hours + 6
    print(f"{hours} hours: {count:.1f} counts per minute")

print(f"Safe to store after {hours} hours")
```

**Why `while`:** you do not know in advance how many halvings it will take, so a `for` loop with a fixed count does not fit. The condition is `>= 50`, not `> 50`: at 30 hours the count is exactly 50, which is not yet **below** 50, so it must halve once more.

### Core 3 (extension): Sports day sprint speeds

```python
runners = ["Ada", "Ben", "Cy", "Dev"]
times = [12.8, 13.4, 12.1, 14.0]

fastest_speed = 0
fastest_name = ""

for i in range(len(runners)):
    speed = 100 / times[i]
    print(f"{runners[i]}: {speed:.2f} m/s")
    if speed > fastest_speed:
        fastest_speed = speed
        fastest_name = runners[i]

print(f"Fastest: {fastest_name} at {fastest_speed:.2f} m/s")
```

**Why:** this is the "best so far" pattern. Start `fastest_speed` at 0 so any real speed beats it, then replace it whenever a faster runner turns up. The fastest runner has the **shortest** time, which is why you compare speeds, not times.

## Fast Track

### Fast Track 1: Linear search

```python
tmax_2025 = [6.8, 8.4, 13.2, 17.3, 19.6, 24.0, 25.3, 24.2, 19.2, 14.6, 12.0, 9.6]


def linear_search(items, target):
    """Return the position of target in items and the number of comparisons made."""
    comparisons = 0
    for position in range(len(items)):
        comparisons = comparisons + 1
        if items[position] == target:
            return position, comparisons
    return -1, comparisons


print(linear_search(tmax_2025, 24.0))
print(linear_search(tmax_2025, 30.0))
```

**Why:** `return` inside the loop stops the search the moment the target is found. The `return -1` line after the loop only runs if every item has been checked, which is why a missing value costs all 12 comparisons.

### Fast Track 2: Binary search

```python
sorted_temps = sorted(tmax_2025)


def binary_search(items, target):
    """Return the position of target in a sorted list and the number of comparisons made."""
    low = 0
    high = len(items) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons = comparisons + 1
        if items[mid] == target:
            return mid, comparisons
        if items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons


print(binary_search(sorted_temps, 24.0))
print(binary_search(sorted_temps, 30.0))

big = list(range(100_000))
print("Linear:", linear_search(big, 100_000))
print("Binary:", binary_search(big, 100_000))
```

**Why:** each comparison throws away half of what is left. `low = mid + 1` and `high = mid - 1` matter: without the `+ 1` and `- 1` the range can stop shrinking and the loop never ends.

Searching 100,000 items for a missing value: linear needs 100,000 comparisons, binary needs 17. Doubling the list adds only one more comparison to binary search. That is why sorted data is worth the effort.

### Fast Track 3 (extension): Square roots by bisection

```python
import math


def square_root(n):
    """Estimate the square root of n by bisection. Return the estimate and the number of halvings."""
    low = 0
    high = max(1, n)
    steps = 0
    while high - low > 0.0001:
        mid = (low + high) / 2
        steps = steps + 1
        if mid * mid < n:
            low = mid
        else:
            high = mid
    return (low + high) / 2, steps


for n in [2, 1000]:
    estimate, steps = square_root(n)
    print(f"square_root({n}) = {estimate:.4f} after {steps} halvings, math.sqrt gives {math.sqrt(n):.4f}")
```

**Why `max(1, n)`:** for numbers below 1 the square root is **bigger** than the number itself. The square root of 0.25 is 0.5. Starting `high` at `n` would miss it, so `high` starts at 1 at least.

Bigger numbers need more halvings because the starting gap is wider: 0 to 1000 needs 24, 0 to 2 needs 15.

## Plenary and exit ticket

**Plenary:** halving shrinks things fast because each step multiplies by a half, so after *k* steps you have 1/2*ᵏ* left. 1600 needs **six** halvings to drop below 50: 800, 400, 200, 100, 50, 25.

**Exit ticket model answer:** use `while` when you do not know in advance how many times the loop must run, and keep going until a condition changes, like the count dropping below 50. Use `for` when you are going through a known collection or a fixed number of steps.
