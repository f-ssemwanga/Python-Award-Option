### Concept 1: Defining a function

A function is a named block of code you can run whenever you want. `def` defines it, the brackets hold its **parameters**, and the indented lines are its body. Nothing happens until you **call** it.

**Predict what prints, then run it.**

```python
def welcome(house):
    print(f"Welcome to {house} House")
    print("-" * 20)


welcome("Rowe")
welcome("Tee")
```

### Concept 2: Returning a value

`print` shows a value on the screen and then it is gone. `return` hands the value **back** to the code that called the function, so it can be stored and used again.

```python
def total_points(academic, sport, music):
    return academic + sport + music


rowe = total_points(412, 358, 97)
tee = total_points(390, 401, 120)

print(f"Rowe {rowe}, Tee {tee}")
print(f"Gap: {abs(rowe - tee)}")
```

**Now break it.** Change `return` to `print`. The totals still appear, but the next line crashes: the function now gives back `None`, and you cannot subtract `None`. Put `return` back.

**Rule of thumb:** a function should usually `return` its answer and leave printing to whoever called it.

### Concept 3: The `math` module

A **module** is a file of ready-made functions. `import` brings one in. Then you use `module.function()`.

| Code | Gives |
| --- | --- |
| `math.sqrt(49)` | `7.0` |
| `math.pi` | `3.14159...` |
| `math.floor(7.9)` | `7` |
| `math.ceil(7.1)` | `8` |

```python
import math

radius = 6.5
area = math.pi * radius ** 2
print(f"A circle of radius {radius} m has area {area:.1f} square metres")

students = 130
per_minibus = 16
print(f"{students} students need {math.ceil(students / per_minibus)} minibuses")
```

**Now break it.** Change `math.ceil` to `math.floor`. You get 8 minibuses, and two students are left behind. Picking the right rounding is part of the logic.
