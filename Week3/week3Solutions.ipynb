## Core

### Core 1: Percentage error

```python
def percentage_error(measured, accepted):
    return abs(measured - accepted) / accepted * 100


error = percentage_error(9.6, 9.81)
print(f"Measured 9.6, accepted 9.81: {error:.1f}% error")
```

**Why:** the function **returns** the number and the caller decides how to print it. That way the same function works for any experiment, and you can use the answer in another calculation. `abs()` drops the minus sign, so reading too low and reading too high both come out as a positive error.

### Core 2: Kinetic energy

```python
def kinetic_energy(mass, speed):
    return 0.5 * mass * speed ** 2


print(f"Rugby player (85 kg at 7.5 m/s): {kinetic_energy(85, 7.5):.1f} J")
print(f"Cricket ball (0.16 kg at 35 m/s): {kinetic_energy(0.16, 35):.1f} J")
```

**Why:** one function, called twice with different values. You write the formula once, so there is only one place it can be wrong.

The ball is faster, but the player is more than 500 times heavier. Speed is squared, so being about 4.7 times faster gives the ball about 22 times more energy per kilogram. That is nowhere near enough to make up for a mass 500 times smaller.

### Core 3 (extension): Is the dice fair?

```python
import random


def count_sixes(rolls):
    sixes = 0
    for roll in range(rolls):
        if random.randint(1, 6) == 6:
            sixes = sixes + 1
    return sixes


for rolls in [60, 600, 60000]:
    print(f"{rolls} rolls: {count_sixes(rolls)} sixes, expected {rolls // 6}")
```

**Why:** the loop variable `roll` is never used, it only counts the repeats. `random.randint(1, 6)` includes both 1 and 6.

Run it a few times. The gap in **sixes** can get bigger as the experiment grows, but the gap as a **percentage** shrinks. With 60 rolls being out by 20% is normal. With 60,000 rolls you are rarely out by more than 1 or 2%. That is why experiments are repeated many times.

## Fast Track

### Fast Track 1: Default arguments

```python
import math


def projectile_range(speed, angle=45, g=9.81):
    return speed ** 2 * math.sin(math.radians(2 * angle)) / g


print(round(projectile_range(20), 2))
print(round(projectile_range(20, 30), 2))
print(round(projectile_range(20, g=1.62), 2))
```

**Why:** parameters with defaults must come after the ones without, so `speed` goes first. In the third call `g=1.62` is a **keyword argument**: it names `g` directly, so `angle` keeps its default of 45. Without the keyword, `projectile_range(20, 1.62)` would set the **angle** to 1.62 degrees.

Convert `2 * angle` to radians, not just `angle`. `math.sin` always works in radians.

### Fast Track 3 (extension): Estimating π with random numbers

```python
import random


def estimate_pi(throws, seed=None):
    random.seed(seed)
    hits = 0
    for throw in range(throws):
        x = random.random()
        y = random.random()
        if x * x + y * y <= 1:
            hits = hits + 1
    return 4 * hits / throws


for throws in [1000, 100_000, 1_000_000]:
    print(f"{throws:>9} throws: {estimate_pi(throws):.5f}")

print(estimate_pi(100_000, seed=2026))
print(estimate_pi(100_000, seed=2026))
```

**Why:** a dart lands inside the quarter circle when its distance from the corner is at most 1. That distance is the square root of x² + y², but comparing x² + y² with 1 gives the same answer without the square root.

With a seed, the "random" numbers are the same sequence every time, so the two seeded lines match exactly. With `seed=None`, Python seeds from the operating system's own source of randomness (or from the clock if there is none), so you get a new answer each run.

More throws give a better estimate, but slowly: to get one more correct digit you need about 100 times as many throws.
