### Concept 1: Lists and positions

A list holds many values in order, inside square brackets. Each value has a position, called an **index**, and **counting starts at 0**.

| Code | Gives |
| --- | --- |
| `houses[0]` | the first item |
| `houses[-1]` | the last item |
| `len(houses)` | how many items |

**Predict what prints, then run it.**

```python
houses = ["Alliott", "Benson", "Collett", "Hayward", "Robert Pearce",
          "Rowe", "Sutton", "Tee", "Trotman", "Young"]

print(houses[0])
print(houses[3])
print(houses[-1])
print(f"There are {len(houses)} houses")
```

**Now break it.** Change `houses[3]` to `houses[10]`. There are 10 houses, but you get an `IndexError`: with counting from 0, the last position is 9.

### Concept 2: Changing a list

Lists can change after you make them.

| Code | Does |
| --- | --- |
| `names.append("Zara")` | adds to the end |
| `names.remove("Ben")` | removes that value |
| `names.sort()` | puts the list in order |
| `"Ada" in names` | `True` if it is there |

```python
team = ["Priya", "Tom", "Ada"]
team.append("Kofi")
team.remove("Tom")
team.sort()

print(team)
print("Ada" in team)
print("Tom" in team)
```

**Now break it.** Run `team.remove("Tom")` a second time. You get a `ValueError`: you cannot remove something that is not there. Check with `in` first when you are not sure.

### Concept 3: Looping over a list

A `for` loop visits each item in turn. Python also has `sum`, `max` and `min` built in.

```python
scores = [72, 58, 91, 66, 84]

total = 0
for score in scores:
    total = total + score

print(f"Total {total}, which matches sum(): {sum(scores)}")
print(f"Highest {max(scores)}, lowest {min(scores)}")
print(f"Mean {total / len(scores):.1f}")
```

### Concept 4: Tuples

A **tuple** uses round brackets. It is like a list, but it **cannot be changed** once made. Use one for facts that belong together and should stay together, like a house and its points.

You can unpack a tuple straight into variables, even inside a `for` loop.

```python
results = [(867, "Rowe"), (911, "Tee"), (802, "Young")]

for points, house in results:
    print(f"{house}: {points}")

winner = max(results)
print(f"Winner: {winner[1]} with {winner[0]}")
```

`max` compares tuples by their first item, which is why the points come first.

**Now break it.** Add `results[0][0] = 1000` at the end. You get a `TypeError`: a tuple cannot be changed. That is the point of using one.
