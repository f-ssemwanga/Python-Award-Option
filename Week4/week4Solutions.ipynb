## Core

### Core 1: Practical kit list

```python
kit = ["beaker", "bunsen burner", "measuring cylinder", "water bath"]

kit.append("thermometer")
kit.append("stopwatch")
kit.remove("bunsen burner")
kit.sort()

for i in range(len(kit)):
    print(f"{i + 1}. {kit[i]}")
print(f"Items: {len(kit)}")
```

**Why:** change the list first, then print it. `sort()` changes the list itself and gives back nothing, so write `kit.sort()` on its own line, not `kit = kit.sort()`. That second version replaces your list with `None`.

`i + 1` is only for the person reading it. Python still counts from 0.

### Core 2: Repeat readings

```python
readings = [15.2, 15.4, 15.1, 16.9, 15.3, 15.2]

mean = sum(readings) / len(readings)
spread = max(readings) - min(readings)
uncertainty = spread / 2
print(f"Mean {mean:.2f} s, range {spread:.2f} s, uncertainty {uncertainty:.2f} s")

readings.remove(16.9)
print("Removed anomaly: 16.9")

mean = sum(readings) / len(readings)
spread = max(readings) - min(readings)
uncertainty = spread / 2
print(f"Mean {mean:.2f} s, range {spread:.2f} s, uncertainty {uncertainty:.2f} s")
```

**Why:** 16.9 is the anomaly because every other reading is within 0.3 s of the rest. Do not call the variable `range`: that is the name of Python's built-in `range()`, and you would break every `for` loop that uses it afterwards.

The three calculation lines appear twice. That works today, and next week's answer would be a function.

### Core 3 (extension): Measuring g with a pendulum

```python
import math

results = [(0.20, 0.90), (0.40, 1.27), (0.60, 1.55), (0.80, 1.80), (1.00, 2.01)]

total = 0
for length, period in results:
    g = 4 * math.pi ** 2 * length / period ** 2
    print(f"L = {length:.2f} m, T = {period:.2f} s, g = {g:.2f}")
    total = total + g

print(f"Mean g = {total / len(results):.2f} m/s²")
```

**Why:** `for length, period in results` unpacks each tuple into two named variables, which reads far better than `result[0]` and `result[1]`. The results are tuples because each length and its period belong together and should never be changed separately.

All five values are within 0.1 of 9.81, so this was a careful experiment.

## Fast Track

### Fast Track 1: Molar mass calculator

```python
ATOMIC_MASS = {"H": 1.008, "C": 12.011, "N": 14.007, "O": 15.999,
               "Na": 22.990, "Cl": 35.45, "Ca": 40.078}


def molar_mass(formula):
    total = 0
    for element, count in formula.items():
        total = total + ATOMIC_MASS[element] * count
    return round(total, 2)


print(molar_mass({"H": 2, "O": 1}))
print(molar_mass({"C": 1, "O": 2}))
print(molar_mass({"C": 6, "H": 12, "O": 6}))
```

**Why:** `.items()` gives each key and its value together, so there is no second lookup. Round only at the end. Rounding each element's share as you go adds small errors that build up.

An element missing from `ATOMIC_MASS`, such as `"Fe"`, raises a `KeyError`. Week 5 is about catching errors like that.

### Fast Track 2: A stack that checks brackets

```python
def balanced(text):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for ch in text:
        if ch in "([{":
            stack.append(ch)
        elif ch in ")]}":
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack


print(balanced("Ca(OH)2"))
print(balanced("[Cu(NH3)4]SO4"))
print(balanced("[Cu(NH3]4)SO4"))
print(balanced("((a+b)*c"))
```

**Why:** the most recent bracket opened must be the first one closed, which is exactly last in, first out. The dictionary maps each closing bracket to its partner, so one test covers all three kinds.

Two checks people miss. `not stack` before `pop()` stops a crash on a string like `")("`. `return not stack` at the end catches brackets left open, as in `"((a+b)*c"`. Returning `True` there would get the last example wrong.

### Fast Track 3 (extension): The lunch queue

```python
from collections import deque

events = ["join Ada", "join Ben", "serve", "join Cy", "serve",
          "join Dev", "serve", "join Eve"]

queue = deque()
for event in events:
    parts = event.split()
    if parts[0] == "join":
        queue.append(parts[1])
    elif parts[0] == "serve":
        print(f"Served: {queue.popleft()}")

print(f"Still waiting: {list(queue)}")
```

**Why:** `deque` takes from the front in one step. `pop(0)` on a list has to shuffle every remaining item along by one, which gets slow when the queue is long. `list(queue)` is only there to make the printout look like a normal list.

A `serve` on an empty queue would raise an `IndexError`. A real program would check `if queue:` first.

## Plenary and exit ticket

**Plenary:** use a **list** when the items are in order and may change, and a **tuple** for a fixed group of facts that belong together. The pendulum results were tuples because a length and its period are one measurement and should not be edited apart. Fast Track: a **dictionary** when you look things up by name, a **stack** when the last thing in must come out first, a **queue** when the first thing in must come out first.
