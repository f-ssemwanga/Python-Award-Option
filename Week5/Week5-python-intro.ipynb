### Concept 1: try and except

When Python cannot do something, it stops with an **error**, also called an **exception**. Wrap risky code in `try`, and say what to do instead in `except`. The program carries on.

**Predict what prints, then run it.**

```python
reading = "21.4"

try:
    temperature = float(reading)
    print(f"Recorded {temperature} C")
except ValueError:
    print(f"'{reading}' is not a number, so it was skipped")

print("Program still running")
```

**Now break it.** Change `reading` to `"21.4C"`, the kind of thing a sensor really sends. Run it: the `except` part runs instead. Then change `except ValueError` to `except ZeroDivisionError` and run it again. Now the error is not the one you are catching, so the program crashes. Put `ValueError` back.

### Concept 2: Keep asking until the input is valid

A **validation loop** asks again and again until the answer is usable. Two checks are needed:

| Check | Catches |
| --- | --- |
| `try` / `except ValueError` | text that is not a number at all, like `ten` |
| `if` | a number that makes no sense, like `-3` repeats |

```python
while True:
    try:
        repeats = int(input("How many repeat readings (1 to 10)? "))
        if 1 <= repeats <= 10:
            break
        print("Between 1 and 10, please")
    except ValueError:
        print("Whole numbers only, please")

print(f"Taking {repeats} readings")
```

Try typing `ten`, then `0`, then `4`.

**Now break it.** Delete the `if` line and the `print` under it, then type `-3`. The loop accepts it. `try` only stops things that are **not numbers**. Checking the number **makes sense** is still your job.

### Concept 3: Cleaning a list of readings

Real data logs have gaps and faults. This list is a day of temperature readings with two bad entries. Skip the bad ones, count them, and use the rest.

```python
log = ["21.4", "21.9", "n/a", "22.1", "", "21.7"]

good = []
bad = 0
for entry in log:
    try:
        good.append(float(entry))
    except ValueError:
        bad = bad + 1

print(f"{len(good)} good readings, {bad} skipped")
print(f"Mean {sum(good) / len(good):.2f} C")
```

**Now break it.** Replace the whole `try` and `except` with just `good.append(float(entry))`. One bad entry and nothing is worked out at all. That is why scientists clean data before using it.
