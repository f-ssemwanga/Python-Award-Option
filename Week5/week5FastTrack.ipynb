### Before Fast Track 1: how `raise` works

`raise` is like `return`, but for a problem. It stops the function on the spot: nothing after it runs and no answer comes back. Instead the error travels back to whoever called the function, and they decide what to do about it.

**Predict what prints, then run it.**

````python
def kinetic_energy(mass, speed):
    """Return kinetic energy in joules."""
    if mass <= 0:
        raise ValueError("mass must be more than zero")
    return 0.5 * mass * speed ** 2


try:
    print(kinetic_energy(85, 7.5))
    print(kinetic_energy(-2, 7.5))
    print("This line never runs")
except ValueError as error:
    print(f"Could not work it out: {error}")
````

`as error` gives you the error itself, and printing it shows the message you wrote after `raise`.

**Why not just print a message inside the function?** Because then the function decides what happens, and it still has to return *something*, usually `None`, which quietly breaks a calculation later on. With `raise`, the function only says what is wrong. The caller chooses: print a message, ask again, or skip that sample. The same function then works in a notebook, a program or a module.

### Before Fast Track 2: `else` and `finally`

`try` has two more optional parts.

| Part | When it runs |
| --- | --- |
| `try` | always tried |
| `except` | only if that error happened |
| `else` | only if **no** error happened |
| `finally` | **always**, error or not |

**Predict which lines print for each value, then run it.**

````python
for text in ["9.81", "abc"]:
    print(f"--- {text}")
    try:
        value = float(text)
        print("try: converted")
    except ValueError:
        print("except: not a number")
    else:
        print("else: no error, so use the value")
    finally:
        print("finally: runs every time")
````

**Why `else`?** It keeps the `try` block down to the one line that can fail. Anything after that goes in `else`, so you never catch an error from a line you were not expecting to fail and mistake it for bad data.

**Why `finally`?** It runs even when something leaves the block early: a `continue`, a `return`, or an error you did not catch. Code placed after the whole `try` block would be skipped in those cases. In real programs `finally` is where you close a file or switch off equipment, whatever happens. You will need that in Week 6.
