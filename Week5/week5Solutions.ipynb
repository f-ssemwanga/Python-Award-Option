## Core

### Core 1: Resistance calculator

```python
try:
    voltage = float(input("Voltage in volts: "))
    current = float(input("Current in amps: "))
    resistance = voltage / current
    print(f"Resistance: {resistance:.1f} ohms")
except ValueError:
    print("Please type numbers only")
except ZeroDivisionError:
    print("The current cannot be zero")
```

**Why:** one `try` can have more than one `except`. Python runs the one that matches the error, so typing `twelve` and typing `0` for the current get different messages.

Use `float`, not `int`. A current of `0.5` amps is normal, and `int("0.5")` is a `ValueError`, so a perfectly good reading would be rejected.

### Core 2: pH reading

```python
while True:
    try:
        ph = float(input("pH reading (0 to 14): "))
        if 0 <= ph <= 14:
            break
        print("pH must be between 0 and 14")
    except ValueError:
        print("Numbers only, please")

print(f"Recorded pH {ph}")
```

**Why:** two different checks for two different problems. `seven` cannot be turned into a number at all, so only `try` can catch it. `15` is a perfectly good number that cannot be a real pH, so only an `if` can catch it.

`while True` with `break` keeps asking until the reading is valid. `0 <= ph <= 14` checks both ends in one go.

### Core 3 (extension): Faulty data logger

```python
logger = ["13.2", "13.8", "ERR", "14.5", "-999", "15.1", "", "14.9", "14.1", "-999"]

good = []
rejected = 0
for entry in logger:
    try:
        value = float(entry)
        if -50 <= value <= 60:
            good.append(value)
        else:
            rejected = rejected + 1
    except ValueError:
        rejected = rejected + 1

print(f"Good readings: {len(good)}")
print(f"Rejected: {rejected}")
print(f"Mean temperature: {sum(good) / len(good):.2f} C")
```

**Why:** `ERR` and the empty string both fail `float()`, so `except` catches them. `-999` does not fail: it is a real number, just not a real temperature. Only the range check stops it.

Leave out the range check and the mean comes out at about -239 C. The program runs without any error and gives a wrong answer, which is worse than crashing.

## Fast Track

### Fast Track 1: Raise your own error

```python
def concentration(moles, volume):
    """Return concentration in mol/dm3."""
    if volume <= 0:
        raise ValueError("volume must be more than zero")
    if moles < 0:
        raise ValueError("moles cannot be negative")
    return moles / volume


samples = [(0.5, 0.25), (0.1, 0), (-0.2, 1.0)]

for moles, volume in samples:
    try:
        print(f"{moles} mol in {volume} dm3: {concentration(moles, volume)} mol/dm3")
    except ValueError as error:
        print(f"{moles} mol in {volume} dm3: {error}")
```

**Why:** the function only decides what is wrong. The loop decides what to do about it, here print the message and carry on to the next sample.

Test `volume <= 0`, not `volume == 0`. A negative volume is just as impossible, and dividing by it would quietly give a negative concentration.

### Fast Track 2: else and finally

```python
readings = ["9.79", "9.8l", "9.83", "nine", "9.81"]
accepted = []
checked = 0

for text in readings:
    try:
        value = float(text)
    except ValueError:
        print(f"Rejected: '{text}'")
    else:
        accepted.append(value)
    finally:
        checked = checked + 1

print(f"Checked {checked}, accepted {accepted}")
```

**Why:** only `float(text)` can fail, so it is the only line inside `try`. The `append` goes in `else`, so it only runs when the conversion worked.

The counting goes in `finally` because `finally` runs even if a `continue` is later added inside `except`. A count placed after the whole `try` block would be skipped for every rejected reading, and `checked` would come out as 3.

### Fast Track 3 (extension): A robust file parser

```python
lines = ["2025,1,62.1", "2025,2,41.0", "2025,3", "2025,4,38.6", "2025,5,4l.2", "2025,6,45.7"]

good = 0
total = 0
line_number = 0
for line in lines:
    line_number = line_number + 1
    parts = line.split(",")
    try:
        rain = float(parts[2])
    except IndexError:
        print(f"Line {line_number}: missing a field")
    except ValueError:
        print(f"Line {line_number}: not a number")
    else:
        good = good + 1
        total = total + rain

print(f"Good lines: {good}, total rain {total:.1f} mm")
```

**Why:** `"2025,3".split(",")` gives only two parts, so `parts[2]` is past the end: `IndexError`. `"4l.2"` has a letter l where the 1 should be, so `float` fails: `ValueError`. Two different errors, two different messages, and the person fixing the file knows which line to look at.

Print the total with `:.1f`. Adding decimals in a computer can give `187.39999999999998`, and rounding only at the end keeps the answer honest.

This is exactly the shape of reading a real data file, which is Week 6.

## Plenary and exit ticket

**Plenary:** use an **`if`** when you can check the value before you use it: is the pH between 0 and 14, is the temperature in a sensible range. Use **`try`** when the only way to find out is to attempt it: turning `"seven"` into a number, dividing by a current of zero. The data logger needed both, because `-999` passes `float()` but fails the range check.

**Exit ticket model answer:**

```python
try:
    mass = float("3.5kg")
    print(mass)
except ValueError:
    print("Not a number")
```

`float` cannot handle the `kg`, so it raises a `ValueError`, and `except` prints the message instead of crashing. `"3.5kg".replace("kg", "")` first would make it work, but that was not the question.
