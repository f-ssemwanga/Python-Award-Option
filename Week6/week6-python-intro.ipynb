### Concept 1: Writing and reading a text file

`open()` connects your program to a file. Use it inside `with`, which closes the file for you when the block ends, even if something goes wrong.

| Mode | Does |
| --- | --- |
| `"w"` | write: creates the file, or **wipes** it if it exists |
| `"a"` | append: adds to the end |
| `"r"` | read: the default |

`\n` is a new line. `write()` does not add one for you.

**Predict what prints, then run it.**

```python
with open("lab_notes.txt", "w") as file:
    file.write("Pendulum practical\n")
    file.write("Length 0.50 m, period 1.42 s\n")

with open("lab_notes.txt", "r") as file:
    print(file.read())
```

Look in the Week 6 folder: `lab_notes.txt` is there. It stays after the program ends, which is the whole point of a file.

### Concept 2: Reading the weather data

The `data` folder holds real Met Office records from five UK stations. It is a **CSV** file: one row per month, values separated by commas.

| Position | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| --- | --- | --- | --- | --- | --- | --- | --- | --- |
| Column | station | year | month | tmax | tmin | af | rain | sun |

`tmax` and `tmin` are the mean daily maximum and minimum temperature in °C, `af` is days of air frost, `rain` is millimetres and `sun` is hours.

The `csv` module splits each line into a list for you. **Everything comes out as a string**, so convert before doing maths.

```python
import csv

oxford_2025 = []
with open(data_file("uk_weather.csv")) as file:
    reader = csv.reader(file)
    next(reader)                      # skip the header row
    for row in reader:
        if row[0] == "Oxford" and row[1] == "2025":
            oxford_2025.append(row)

print(f"{len(oxford_2025)} months found")
print(oxford_2025[6])

july_rain = float(oxford_2025[6][6])
print(f"Rain in July 2025: {july_rain} mm")
```
