### Concept 1: A class and its objects

A **class** is a design for a new kind of thing. An **object** is one thing made from that design. `__init__` runs when an object is made, and stores its data as **attributes** on `self`, which means *this object*.

```python
class Sample:
    def __init__(self, name, mass, volume):
        self.name = name
        self.mass = mass          # grams
        self.volume = volume      # cm3


copper = Sample("Copper", 89.6, 10.0)
granite = Sample("Granite", 54.0, 20.0)

print(copper.name, copper.mass)
print(granite.name, granite.mass)
```

Two objects, one design. Each keeps its own values.

### Concept 2: Methods

A **method** is a function that belongs to a class. Its first parameter is always `self`, so it can use the object's own attributes.

```python
class Sample:
    def __init__(self, name, mass, volume):
        self.name = name
        self.mass = mass
        self.volume = volume

    def density(self):
        return self.mass / self.volume

    def describe(self):
        return f"{self.name}: {self.density():.2f} g/cm3"


samples = [Sample("Copper", 89.6, 10.0), Sample("Granite", 54.0, 20.0), Sample("Ice", 9.2, 10.0)]

for sample in samples:
    print(sample.describe())
```

A list of objects works like any other list. Each object knows how to describe itself, so the loop stays short.

### Concept 3: A weather station that charts itself

The data, and the things you do with it, now live together. This class reads its own readings from the CSV and draws its own chart.

```python
import csv
import matplotlib.pyplot as plt


class WeatherStation:
    def __init__(self, name):
        self.name = name
        self.months = []
        self.tmax = []

    def load(self, year):
        with open(data_file("uk_weather.csv")) as file:
            reader = csv.reader(file)
            next(reader)
            for row in reader:
                if row[0] == self.name and row[1] == str(year):
                    self.months.append(int(row[2]))
                    self.tmax.append(float(row[3]))

    def plot(self):
        plt.plot(self.months, self.tmax, marker="o", label=self.name)


for name in ["Lerwick", "Oxford", "Camborne"]:
    station = WeatherStation(name)
    station.load(2025)
    station.plot()

plt.xlabel("Month")
plt.ylabel("Mean daily maximum (°C)")
plt.title("Three stations, 2025")
plt.legend()
plt.show()
```

Three stations, and the loop is three lines long. Adding a fourth is one word.

**Now break it.** Swap `str(year)` for just `year`. Nothing errors, but nothing is drawn either: `"2025" == 2025` is `False`. Strings from a file never equal numbers.
