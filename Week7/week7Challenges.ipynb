A class can be built on another one. The new class **inherits** every attribute and method, and adds its own.

`Sensor` is written for you. Write `TemperatureSensor(Sensor)`, which:

- calls the parent's `__init__` with `super().__init__(name)`, then stores `units = "C"`
- adds a method `mean_fahrenheit()` that converts the mean to Fahrenheit

```text

Readings: 5
```

`TemperatureSensor` never defines `add` or `mean`, yet it can use both.

````python
# Stuck on Challenge 3? Have a proper go first, then run this cell for a hint.
import sys; sys.path.append(".."); from coursetools import hint
hint(7, "Challenge 3")
````

````python
class Sensor:
    def __init__(self, name):
        self.name = name
        self.readings = []

    def add(self, value):
        self.readings.append(value)

    def mean(self):
        return sum(self.readings) / len(self.readings)


# Challenge 3: your code below this line
````
